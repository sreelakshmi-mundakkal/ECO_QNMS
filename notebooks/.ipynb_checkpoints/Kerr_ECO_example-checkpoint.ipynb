```julia
import ApproxFun 

include("../scripts/TeukolskyQNMFunctions.jl")
import .TeukolskyQNMFunctions as QNM
```

```julia
nr=100  # number of radial points
nl=100  # number of angular points
s=-2   # field spin
m=2    # m angular value
l=2    # l angular value
n=0    # overtone number
a=1e-6  # dimensionless black hole spin 
eps =0.001
qnm_tolerance=1e-8
coef_tolerance=1e-6
epsilon=1e-6
T=Float64

## QNM value near n=0,l=2,m=2, s=-2 for dimensionless spin a=0.7

omega_0= 0.5456404087339959 - 0.0006588774318067*im   # Shwarzschild ECO known value 
```

```julia
omega, lambda, vs, vr, rs = QNM.compute_omdS(
            nr, nl, s, l, m, a,eps, omega_0,
            tolerance=qnm_tolerance,
            epsilon=epsilon,
            gamma=1.0-a,
            verbose=false
         );
println(omega)
println(lambda)
```
